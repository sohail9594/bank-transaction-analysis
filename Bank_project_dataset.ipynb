```python
import pandas as pd
import numpy as np
```

```python
np.random.seed(42)

n = 1000

df = pd.DataFrame({
    "Customer_ID": np.random.randint(1000, 1100, n),
    "Transaction_Date": pd.date_range(start="2023-01-01", periods=n, freq="D"),
    "Transaction_Amount": np.random.randint(100, 10000, n),
    "Transaction_Type": np.random.choice(["Deposit", "Withdrawal"], n)
})

df.head()
```

```python
df.dropna(inplace=True)
df['Transaction_Date'] = pd.to_datetime(df['Transaction_Date'])
```

```python
df['Month'] = df['Transaction_Date'].dt.to_period('M')
```

```python
df.to_csv("bank_data_cleaned.csv", index=False)
```
